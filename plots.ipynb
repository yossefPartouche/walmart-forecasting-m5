```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python


# 1. Load your data (replace with your actual file paths)
# df_train = pd.read_csv('training_data.csv')
# df_events = pd.read_csv('events.csv')
df_train    = pd.read_csv('data/train.csv')           # columns: store_id, date, revenue
df_total = df_train[df_train['store_name'] == 'All Stores'].copy()
df_total = df_total.drop(columns=['store_id', 'store_name'])


df_events = pd.read_csv('data/calendar_events.csv') # columns: date, event_name (etc.)

# Ensure dates are datetime objects for proper Plotly formatting
df_train['date'] = pd.to_datetime(df_train['date'])
df_events['date'] = pd.to_datetime(df_events['date'])

# 2. Create the base line chart for store revenues
# 'color' automatically groups the data by store_name and assigns different colors
fig = px.line(
    df_train, 
    x='date', 
    y='revenue', 
    color='store_name',
    title='Store Revenue Over Time with Key Events',
    labels={'date': 'Date', 'revenue': 'Revenue ($)', 'store_name': 'Store'}
)

# 3. Add a vertical line for each event in the events dataset
for index, row in df_events.iterrows():
    fig.add_vline(
        x=row['date'].timestamp() * 1000, # Plotly expects timestamp in ms for vlines on datetime axes sometimes, but usually just the date string/object works. Using row['date'] directly is usually fine in newer versions.
        # simpler alternative: x=row['date'],
        line_width=2, 
        line_dash="dash", 
        line_color="red",
        annotation_text=row['event'], 
        annotation_position="top right", # Places the event name at the top of the line
        annotation_textangle=-90 # Optional: turns the text sideways if you have many events
    )

# 4. Improve layout for better readability
fig.update_layout(
    hovermode="x unified", # Shows tooltip for all stores at a specific date when hovering
    template="plotly_white"
)

# 5. Display the interactive plot
fig.show()
```
